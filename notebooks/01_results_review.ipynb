# Lagos Public Transport Accessibility — Results Review
This notebook reviews the supplied project tables and reproduces key statistics.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
lga = pd.read_csv(ROOT / "data/processed/tables/lga_accessibility_results.csv")
lga.head()
```

```python
summary = {
    "LGA-assigned population": lga.Population.sum(),
    "Within 30 minutes": lga.Within_30min.sum(),
    "Underserved": lga.Underserved.sum(),
    "30-minute share (%)": 100*lga.Within_30min.sum()/lga.Population.sum(),
}
pd.Series(summary)
```
